```python
def write_fastq(forward, reverse, checks, singleton_choice):
    
    """
    Questa funzione scrive le read trimmate che hanno superato il controllo di lunghezza su due nuovi file FastQs. Dà la possibilità, qualora
    l'utente lo voglia, di salvare in un altro file i singleton contandoli.
    ---------------
    Parametri:
    forward, tuple
        Tupla delle quattro righe della read forward
    reverse, tuple
        Tupla delle quattro righe della read reverse
    checks, tuple
        Tupla di due booleani, che descrivono lo stato della coppia di read (se hanno superato il controllo della lunghezza)
    singleton_choice, bool
        Se True, salva i singleton in un altro file; altrimenti, no.
    ---------------
    Returns:
    singleton_counting, tuple
        Tupla di due interi corrispondenti al contatore dei singleton forward e al contatore dei singleton reverse.
    """

    if checks[0] & checks[1]:
        f3.writelines(forward)
        f4.writelines(reverse)

    singleton_forward_count=0
    singleton_reverse_count=0
    
    if singleton_choice:
        match(checks[0], checks[1]):
            case(False, True):
                f5.writelines(reverse)
                singleton_forward_count+=1
            case(True, False):
                f5.writelines(forward)
                singleton_reverse_count+=1
    singleton_counting=(singleton_forward_count, singleton_reverse_count)
    return singleton_counting
```
