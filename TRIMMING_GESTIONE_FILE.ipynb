```python
import os

def acquisisci_file():
    """
    Chiede all'utente i percorsi di R1 e R2 e verifica che esistano, 
    non siano vuoti e non siano lo stesso file. 
    Inoltre indica quale file ha un problema.
    Restituisce: (file_r1, file_r2)
    """
    while True:
        file_r1=input("Inserisci il percorso del file R1 (Forward): ").strip()
        file_r2=input("Inserisci il percorso del file R2 (Reverse): ").strip()

        #Controlla l'esistenza specifica per entrambi
        if not os.path.exist(file_r1):
            print(f"ERRORE:Il file R1 ('{file_r1}') non esiste. Controlla il nome e riprova.\n")

        elif not os.path.exist(file_r2):
            print(f"ERRORE:Il file R2 ('{file_r2}') non esiste. Controlla il nome e riprova.\n")

        #Controlla dimensione (0 byte) specifica per entrambi
        elif os.stat(file_r1).st.size==0:
            print(f"ERRORE: il file R1 ('{file_r1}') è vuoto (0 byte). Riprova.\n")
        elif os.stat(file_r2).st.size==0:
            print(f"ERRORE: il file R2 ('{file_r2}') è vuoto (0 byte). Riprova.\n")

        #Controllo che non sia stato inserito lo stesso file due volte
        elif file_r1==file_r2:
            print(f"ERRORE. Hai inserito due volte lo stesso file ('{file_r1}') sia per R1 che per R2. Riprova.\n")

        else:
            return file_r1, file_r2
        
```

```python
def acquisisci_soglie():
    """
    Chiede all'utente Qmin e Lmin (con default 20 e 50) e verifica che siano numeri interi positivi.
    Restituisce: (q_min, l_min)
    """
    while True:
        try:
            q_input=input("Inserisci la soglia di qualità Qmin [default=20]").strip
            q_min=20 if q_input=="" else int(q_input)

            l_input=input("Inserisci la lunghezza minima Lmin [default=50]").strip
            l_min=50 if l_input=="" else int(l_input)

            if q_min <0 or l_min <=0:
                raise ValueError
            return q_min, l_min

        except ValueError:
            print("ERRORE: Inserisci numeri interi positivi (es. 20 e 50). Riprova. \n")
```

```python
def acquisisci_singleton():
    """
    Chiede all'utente se vuole salvare i singleton (s/n).
    Restituisce: True (se 's') oppure False (se 'n')
    """
    while True:
        scelta=(input("Vuoi salvare le read singleton in un file? (s/n): ").strip().lower())

    if scelta in ("s", "n"):
        return scelta=="s"
    else:
        print("ERRORE: Risposta non valida. Digita 's' per si o 'n' per no.\n")
        
```

```python
def acquisisci_parametri():
    """
    Chiama in sequenza le 3 funzioni di controllo e restituisce
    tutti i dati pronti
    """
    file_r1, file_r2 = acquisisci_file()
    q_min, l_min = acquisisci_soglie()
    salva_singleton = acquisisci_singleton()

    return file_r1, file_r2, q_min, l_min, salva_singleton
```
