```python
def check_Length(forward, reverse, minLen):
    """
    Questa funzione controlla se le due read di una coppia (R1 forward e R2 reverse), dopo il
    trimming, raggiungono la lunghezza minima richiesta.

    Parametri
    ---------
    forward : tupla
    Tupla di 2 stringhe restituita dalla funzione di trimming per la read forward: [sequenza, qualita'].
    reverse : tupla
    Tupla di 2 stringhe restituita dalla funzione di trimming per la read reverse: [sequenza, qualita'].
    minLen : int 
    Lunghezza minima che una read deve avere per essere conservata.

    Funzionamento
    -------------
    Calcola la lunghezza della sequenza (primo elemento di ciascuna lista) di R1 e di R2 e la confronta con 
    minLen. Una read supera il controllo se la sua lunghezza e' maggiore o uguale a minLen.
    Dalla combinazione dei due valori booleani restituiti si ricava lo stato della coppia:
        (True,  True)  -> coppia conservata (entrambe le read superano il controllo)
        (True,  False) -> singleton: si conserva solo R1
        (False, True)  -> singleton: si conserva solo R2
        (False, False) -> coppia scartata

    Restituisce
    -----------
    tuple(bool, bool)
        (R1_passed, R2_passed): True se la read corrispondente e' lunga almeno minLen, False altrimenti.
    """
    # Estraggo la sequenza (elemento 0 della lista) e ne calcolo la lunghezza
    length_R1 = len(forward[0])
    length_R2 = len(reverse[0])

    # Confronto la lunghezza di ciascuna read con la soglia minima
    R1_passed = length_R1 >= minLen
    R2_passed = length_R2 >= minLen
    checks = (R1_passed, R2_passed)

    # Restituisco i due esiti: insieme descrivono lo stato della coppia
    return checks
```
